## Exploratory Business Analysis

This stage examines sales performance, profitability, customer and product segments, geographic performance, and trends over time.

The objective is to identify the primary drivers of business performance and develop evidence-based recommendations.

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order ID"].nunique()
total_quantity = df["Quantity"].sum()
profit_margin = total_profit / total_sales

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Total Orders: {total_orders:,}")
print(f"Units Sold: {total_quantity:,}")
print(f"Overall Profit Margin: {profit_margin:.2%}")
```

```python
category_performance = (
    df.groupby("Category")
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum")
      )
      .sort_values("Sales", ascending=False)
)

category_performance["Profit Margin"] = (
    category_performance["Profit"] / category_performance["Sales"]
)

category_performance
```

```python
subcategory_performance = (
    df.groupby(["Category", "Sub-Category"])
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum")
      )
)

subcategory_performance["Profit Margin"] = (
    subcategory_performance["Profit"] /
    subcategory_performance["Sales"]
)

subcategory_performance = (
    subcategory_performance
    .sort_values("Profit", ascending=True)
)

subcategory_performance
```

```python
discount_analysis = (
    df.groupby("Discount")
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum"),
          Transactions=("Order ID", "count")
      )
      .reset_index()
)

discount_analysis["Profit Margin"] = (
    discount_analysis["Profit"] /
    discount_analysis["Sales"]
)

discount_analysis
```

```python
furniture_discount_analysis = (
    df[df["Category"] == "Furniture"]
    .groupby(["Sub-Category", "Discount"])
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Transactions=("Order ID", "count")
    )
    .reset_index()
)

furniture_discount_analysis["Profit Margin"] = (
    furniture_discount_analysis["Profit"] /
    furniture_discount_analysis["Sales"]
)

furniture_discount_analysis[
    furniture_discount_analysis["Sub-Category"].isin(["Tables", "Bookcases"])
]
```

### Profitability and Discount Finding

Furniture generated substantial sales but significantly weaker profitability than Technology and Office Supplies.

Sub-category analysis identified Tables and Bookcases as major contributors to the profitability weakness. Both generated negative profit overall.

Further analysis identified a strong association between discount levels and profitability:

- Undiscounted Bookcases generated approximately a 19.0% profit margin.
- Undiscounted Tables generated approximately an 18.6% profit margin.
- Bookcase margins became negative at discounts of 30% and above.
- Table margins were approximately break-even at a 20% discount and became increasingly negative at higher discount levels.
- Across the full dataset, every observed discount level of 30% or greater produced a negative aggregate profit margin.

These findings suggest that discount strategy is an important contributor to the profitability problem, particularly for Tables and Bookcases.

**Business implication:** Management should review discount policies for these sub-categories, particularly discounts of 30% or greater, and evaluate whether lower discount thresholds could preserve sales while improving profitability.

This analysis identifies an association between discounting and profitability and does not establish causation.

```python
yearly_performance = (
    df.assign(Year=df["Order Date"].dt.year)
      .groupby("Year")
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum"),
          Orders=("Order ID", "nunique")
      )
)

yearly_performance["Profit Margin"] = (
    yearly_performance["Profit"] /
    yearly_performance["Sales"]
)

yearly_performance
```

### Yearly Performance Finding

Sales, profit, and order volume show substantial overall growth from 2023 through 2026.

Sales increased from approximately $494K in 2023 to $746K in 2026, while orders increased from 995 to 1,723.

Profit margin improved from 10.46% in 2023 to 13.46% in 2025. However, margin declined to 12.87% in 2026 despite 2026 producing the highest sales and profit.

This suggests that revenue growth alone does not fully explain business performance. Further analysis is required to identify the factors contributing to the 2026 margin decline.

```python
category_yearly = (
    df.assign(Year=df["Order Date"].dt.year)
      .groupby(["Year", "Category"])
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum")
      )
      .reset_index()
)

category_yearly["Profit Margin"] = (
    category_yearly["Profit"] /
    category_yearly["Sales"]
)

category_yearly[
    category_yearly["Year"].isin([2025, 2026])
]
```

```python
furniture_detail = (
    df[
        (df["Order Date"].dt.year.isin([2025, 2026])) &
        (df["Category"] == "Furniture")
    ]
    .assign(Year=lambda x: x["Order Date"].dt.year)
    .groupby(["Year", "Sub-Category"])
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum")
    )
    .reset_index()
)

furniture_detail["Profit Margin"] = (
    furniture_detail["Profit"] /
    furniture_detail["Sales"]
)

furniture_detail
```

```python
### 2026 Margin Decline Investigation

The decline in overall profit margin from 13.46% in 2025 to 12.87% in 2026 was not distributed evenly across the business.

Furniture margin declined from 3.59% to 1.53%, while Office Supplies also declined from 19.11% to 16.44%. Technology improved from 17.58% to 18.66%.

Within Furniture, Tables were the clearest source of deterioration. Table sales remained approximately $61K in both years, but losses increased from about $3.0K in 2025 to $8.1K in 2026. Table profit margin consequently declined from -4.85% to -13.27%.

Bookcases also moved from approximately break-even to a negative margin, while Chairs and Furnishings improved.

These results indicate that the Furniture decline was concentrated primarily in Tables rather than reflecting deterioration across all Furniture sub-categories.
```

### 2026 Margin Decline Investigation

The decline in overall profit margin from 13.46% in 2025 to 12.87% in 2026 was not distributed evenly across the business.

Furniture margin declined from 3.59% to 1.53%, while Office Supplies also declined from 19.11% to 16.44%. Technology improved from 17.58% to 18.66%.

Within Furniture, Tables were the clearest source of deterioration. Table sales remained approximately $61K in both years, but losses increased from about $3.0K in 2025 to $8.1K in 2026. Table profit margin consequently declined from -4.85% to -13.27%.

Bookcases also moved from approximately break-even to a negative margin, while Chairs and Furnishings improved.

These results indicate that the Furniture decline was concentrated primarily in Tables rather than reflecting deterioration across all Furniture sub-categories.
